```python
import numpy as np
```

```python
employees=np.array(['arun','bavana','charan','divya','eshan','fathima'])
print("employees=",employees)
print()
performance=np.array(
    ([78,82,85,88],
    [65,70,68,74],
    [92,89,95,91],
    [55,62,60,67],
    [81,79,84,87],
    [72,75,78,80]),

)
print("performance=",performance)
print()

Quarters=np.array(['q1','q2','q3','q4'])
print('Quarters',Quarters)
```

```python
#1
print("total_performance_score =",performance.sum(axis=1))
```

```python
# 2
print("average quarterly performance",performance.mean(axis=1))
```

```python
# 3
print("average performance of each quarter",performance.mean(axis=0))
```

```python
# 4
total=performance.sum(axis=1)
highest=(employees[total.argmax()])
print("highest total performance:",highest)
```

```python
# 5
i=np.argsort(total)
print(employees[i][::-1])
```
